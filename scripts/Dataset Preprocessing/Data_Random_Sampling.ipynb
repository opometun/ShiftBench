```python
import random
import shutil
from pathlib import Path
import math
```

```python
Folder_Image = Path("C:/Users/KRISHNENDU/Desktop/EAIHD DATA/ALL MODEL DATA/New folder (3)") # Folder name of the preprocessed images
Folder_Mask = Path("C:/Users/KRISHNENDU/Desktop/EAIHD DATA/ALL MODEL DATA/New folder (2)") # Folder name of the annotation masks of the aforementioned images

Folder_Sampled_Image = Path("C:/Users/KRISHNENDU/Desktop/EAIHD DATA/ALL MODEL DATA/Folder_Sampled_Image") # Destination folder of the randomly sampled images
Folder_Sampled_Mask = Path("C:/Users/KRISHNENDU/Desktop/EAIHD DATA/ALL MODEL DATA/Folder_Sampled_Mask") # Destination folder of the annotation masks of the aforementioned randomly sampled images

Folder_Sampled_Image.mkdir(parents = True, exist_ok = True) # Create the storage folder of the aforementioned randomly sampled images
Folder_Sampled_Mask.mkdir(parents = True, exist_ok = True) # Create the storage folder of the aforementioned randomly sampled annotation masks

sample_size = input('Please enter the percentage of data that you want to randomly sample: ') # User inputs the required percentage of the data that's needed to be sampled (25 for 25%, 50 for 50%, 75 for 75% and so on)
percent_calculation = int(sample_size)/100 # Converts the user inputted percentage amount to a fraction

Files_Image = {f.name for f in Folder_Image.iterdir() if f.is_file()} # Get the filenames of all the images in the source folder
Files_Mask = {f.name for f in Folder_Mask.iterdir() if f.is_file()} # Get the filenames of all the annotation masks in the source folder

Matching_Files = list(Files_Image.intersection(Files_Mask)) # Make a list of the exactly same filenames in both the aforementioned folders

if (100/len(Matching_Files)) < percent_calculation : # Checks if the length of the matching files list is less than the percentage of data that needs to be sampled
    print(f"Warning: I only found {len(Matching_Files)} matching pairs. I have sampled all of these.")
    Sampled_Filenames = Matching_Files
else:
    percent_sample_size = math.ceil((len(Matching_Files))*percent_calculation) # Converts the percentage quantity to the numeric amount of data that needs to be randomly sampled
    Sampled_Filenames = random.sample(Matching_Files, percent_sample_size) # Samples the amount of data mentioned above

for filename in Sampled_Filenames:
    shutil.copy(Folder_Image / filename, Folder_Sampled_Image / filename) # Copies the sampled image data to its respective target folder
    shutil.copy(Folder_Mask / filename, Folder_Sampled_Mask / filename) # Copies the sampled annotataion masks to its respective target folder

print(f"I have successfully paired and copied {len(Sampled_Filenames)} into the new folders.")
```
