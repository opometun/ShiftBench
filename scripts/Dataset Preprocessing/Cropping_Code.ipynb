```python
from pathlib import Path
from PIL import Image
```

```python
source_dir = Path("C:/Users/KRISHNENDU/Desktop/EAIHD DATA/ALL MODEL DATA/GTA MODEL DATA TRAIN/GTA RESIZED TRAIN DATA/GTA_IMAGE_DATA_TRAIN_LANCZOS_1024_563") # Folder location of the images that need to be cropped
target_dir = Path("C:/Users/KRISHNENDU/Desktop/EAIHD DATA/ALL MODEL DATA/GTA MODEL DATA TRAIN/GTA RESIZED TRAIN DATA/GTA_IMAGE_DATA_TRAIN_LANCZOS_1024_512") # Target folder of the cropped images

crop_box = (1024, 512) # Size of the crop box for implementing center croping

target_dir.mkdir(parents = True, exist_ok = True) # Create the storage folder of the cropped images

valid_extensions = {'.jpg', '.jpeg', '.png', '.bmp', '.webp'} # A dictionary to store all the relevant and useful image extensions that one might encounter

for img_path in source_dir.iterdir(): # Iterate through the source image directory
    if img_path.is_file() and img_path.suffix.lower() in valid_extensions:
        try:
            with Image.open(img_path) as img: # Open each individual image if the image files are present in the directory with all the relevant extensions mentioned above

                left = round((img.size[0] - crop_box[0])/2) # Get the starting index for center cropping box
                top = round((img.size[1] - crop_box[1])/2) # Get the ending index for image cropping box
                cropped_image = img.crop((left, top, crop_box[0] + left, crop_box[1] + top)) # Implement center cropping

                output_path = target_dir / img_path.name # Create the output directory

                cropped_image.save(output_path) # Save the cropped images in the output path

        except Exception as e:
            print(f"I could not process {img_path.name}: {e}") # Raise an exception if the images in the directory can't be accessed

print(f"All images successfully resized and saved to {target_dir}")
```
