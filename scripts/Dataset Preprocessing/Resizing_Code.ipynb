```python
from pathlib import Path
from PIL import Image
```

```python
source_dir = Path("C:/Users/KRISHNENDU/Desktop/EAIHD DATA/ALL MODEL DATA/CityScapes MODEL DATA/CityScapes TEST/CityScapes_TEST_ANNOTATION_MASKS") # Folder location of the images that need to be resized
target_dir = Path("C:/Users/KRISHNENDU/Desktop/EAIHD DATA/ALL MODEL DATA/CityScapes MODEL DATA/CityScapes TEST/CityScapes_TEST_ANNOTATION_MASKS_RESIZED") # Target folder of the resized images
target_size = (1024, 512) # Target resolution of the image

target_dir.mkdir(parents = True, exist_ok = True) # Create the storage folder of the resized images

valid_extensions = {'.jpg', '.jpeg', '.png', '.bmp', '.webp'} # A dictionary to store all the relevant and useful image extensions that one might encounter

for img_path in source_dir.iterdir(): # Iterate through the source image directory
    if img_path.is_file() and img_path.suffix.lower() in valid_extensions:
        try:
            with Image.open(img_path) as img: # Open each individual image if the image files are present in the directory with all the relevant extensions mentioned above
                resized_image = img.resize(target_size, Image.Resampling.LANCZOS) # LANCZOS technique for resizing image data towards amazing upscaling and downscaling quality
                #resized_image = img.resize(target_size, Image.Resampling.NEAREST) # LANCZOS technique for resizing annotation masks towards preserving the exact label mask values

                output_path = target_dir / img_path.name # Create the output directory

                resized_image.save(output_path) # Save the resized images in the output path

        except Exception as e:
            print(f"I could not process {img_path.name}: {e}") # Raise an exception if the images in the directory can't be accessed

print(f"All images successfully resized and saved to {target_dir}")
```
